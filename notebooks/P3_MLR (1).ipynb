```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

```

```python
df = pd.read_csv(r"C:\KLH\2026-2027\Subjects\Y25-ML\practical\Placement-Prediction-System\data\placement_predict_50k Dataset.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
# ============================================================
# 1. SELECT PLACED STUDENTS
# ============================================================

placed_df = df[df["Salary Package"] > 0].copy()

print("Placed students:", len(placed_df))
```

```python
# ============================================================
# 2. DEFINE FEATURES AND TARGET
# ============================================================

feature_cols = [
    "CGPA",
    "AptitudeTestScore",
    "SoftSkillsRating",
    "CodingTestScore",
    "MockInterviewScore"
]

X = placed_df[feature_cols]
y = placed_df["Salary Package"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())
```

```python

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))
```

```python
print("Missing values in X_train:")
print(X_train.isna().sum())

print("\nMissing values in X_test:")
print(X_test.isna().sum())
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)
```

```python
print("Missing values in X_train after imputation:")
print(np.isnan(X_train_imputed).sum())

print("\nMissing values in X_test after imputation:")
print(np.isnan(X_test_imputed).sum())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)
```

```python
from sklearn.linear_model import LinearRegression

mlr_model = LinearRegression()

mlr_model.fit(
    X_train_scaled,
    y_train
)
```

```python
from sklearn.linear_model import LinearRegression

mlr_model = LinearRegression()

mlr_model.fit(
    X_train_scaled,
    y_train
)
```

```python

#Predict on test data
y_pred = mlr_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)
import numpy as np

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

mae = mean_absolute_error(
    y_test,
    y_pred
)

r2 = r2_score(
    y_test,
    y_pred
)

print("Multiple Linear Regression Results")
print("-----------------------------------")
print("RMSE:", rmse)
print("MAE :", mae)
print("R²  :", r2)
```

```python
## Give a new student's CGPA,AptitudeTestScore,SoftSkillsRating,CodingTestScore,MockInterviewScore
new_student = np.array([[8.5, 75, 4.2, 80, 72]])

new_student_imputed = imputer.transform(new_student)

new_student_scaled = scaler.transform(new_student_imputed)

predicted_salary = mlr_model.predict(new_student_scaled)[0]

print(f"Predicted Salary Package: {predicted_salary:.2f} LPA")
```
