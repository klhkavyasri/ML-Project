```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("D:/Placement-Predication-System/data/titanic.csv")
output_dir = "D:/Placement-Predication-System/outputs"
os.makedirs(output_dir, exist_ok=True)
sns.set(style="whitegrid")
```

```python
# 1. Correlation Heatmap
plt.figure(figsize=(10,8))
corr = df.select_dtypes(include=['int64','float64']).corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "01_Correlation_Heatmap.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 2. Missing Values Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(df.isnull(), cbar=False, cmap="viridis")
plt.title("Missing Values Heatmap")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "02_Missing_Values_Heatmap.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 3. Survival Count
plt.figure(figsize=(6,4))
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "03_Survival_Count.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 4. Gender Distribution


plt.figure(figsize=(6,4))
sns.countplot(x='Sex', data=df)
plt.title("Gender Distribution")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "04_Gender_Distribution.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 5. Survival by Gender


plt.figure(figsize=(6,4))
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Survival by Gender")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "05_Survival_by_Gender.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 6. Passenger Class Distribution


plt.figure(figsize=(6,4))
sns.countplot(x='Pclass', data=df)
plt.title("Passenger Class Distribution")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "06_Pclass_Distribution.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 7. Survival by Passenger Class


plt.figure(figsize=(6,4))
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title("Survival by Passenger Class")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "07_Survival_by_Pclass.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 8. Age Distribution


plt.figure(figsize=(8,5))
sns.histplot(df['Age'].dropna(), kde=True, bins=30)
plt.title("Age Distribution")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "08_Age_Distribution.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 9. Fare Distribution
# ============================================

plt.figure(figsize=(8,5))
sns.histplot(df['Fare'], kde=True, bins=30)
plt.title("Fare Distribution")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "09_Fare_Distribution.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 10. Age Boxplot
# ============================================

plt.figure(figsize=(6,4))
sns.boxplot(x=df['Age'])
plt.title("Age Boxplot")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "10_Age_Boxplot.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 11. Fare Boxplot
# ============================================

plt.figure(figsize=(6,4))
sns.boxplot(x=df['Fare'])
plt.title("Fare Boxplot")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "11_Fare_Boxplot.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 12. Age vs Survival
# ============================================

plt.figure(figsize=(6,4))
sns.boxplot(x='Survived', y='Age', data=df)
plt.title("Age vs Survival")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "12_Age_vs_Survival.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 13. Fare vs Survival
# ============================================

plt.figure(figsize=(6,4))
sns.boxplot(x='Survived', y='Fare', data=df)
plt.title("Fare vs Survival")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "13_Fare_vs_Survival.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 14. Embarked Distribution
# ============================================

plt.figure(figsize=(6,4))
sns.countplot(x='Embarked', data=df)
plt.title("Embarked Distribution")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "14_Embarked_Distribution.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 15. Family Size Distribution
# ============================================

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

plt.figure(figsize=(8,5))
sns.histplot(df["FamilySize"], bins=10)
plt.title("Family Size Distribution")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "15_Family_Size_Distribution.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 16. Survival by Family Size
# ============================================

plt.figure(figsize=(8,5))
sns.barplot(x="FamilySize", y="Survived", data=df)
plt.title("Survival by Family Size")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "16_Survival_by_Family_Size.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 17. Pair Plot
# ============================================

pair = sns.pairplot(
    df[['Survived','Pclass','Age','Fare']].dropna(),
    hue='Survived'
)

pair.savefig(os.path.join(output_dir, "17_Pairplot.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 18. Survival Percentage
# ============================================

plt.figure(figsize=(6,6))
plt.pie(
    df["Survived"].value_counts(),
    labels=["Not Survived","Survived"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Survival Percentage")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "18_Survival_Percentage.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 19. Passenger Class Percentage
# ============================================

plt.figure(figsize=(6,6))
plt.pie(
    df["Pclass"].value_counts(),
    labels=df["Pclass"].value_counts().index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Passenger Class Percentage")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "19_Pclass_Percentage.png"), dpi=300)
plt.show()
plt.close()
```

```python
# 20. Age vs Fare Scatter Plot
# ============================================

plt.figure(figsize=(8,6))
sns.scatterplot(
    x='Age',
    y='Fare',
    hue='Survived',
    data=df
)
plt.title("Age vs Fare")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "20_Age_vs_Fare.png"), dpi=300)
plt.show()
plt.close()

print("All EDA graphs have been saved successfully!")
print("Output Folder:", output_dir)
```
